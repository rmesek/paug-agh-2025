```python
# Set the necessary paths
# Choose parameters depending on hardware capabilities
# Check if the data has been downloaded correctly.

input_dir = Path("images/")
target_dir = Path("annotations/trimaps/")
img_size = 128
# Object, background, uncertain area between
num_classes = 3
batch_size = 16

input_img_paths = sorted(
    [fname for fname in input_dir.iterdir() if fname.suffix == ".jpg"]
)
target_img_paths = sorted(
    [target_dir / (fname.stem + ".png") for fname in input_img_paths]
)

print("Number of samples:", len(input_img_paths))
```

```python
# Show an example image and the corresponding mask

plt.figure(figsize=(10, 5))
for i in range(3):
    plt.subplot(2, 3, i + 1)
    img = plt.imread(input_img_paths[i])
    plt.imshow(img)
    plt.axis("off")
    plt.title("Input Image")

    plt.subplot(2, 3, i + 4)
    mask = plt.imread(target_img_paths[i])
    plt.imshow(mask)
    plt.axis("off")
    plt.title("Mask")
plt.show()
```
